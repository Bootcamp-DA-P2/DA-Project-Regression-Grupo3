Investigar nulos anormales

```python
# Ver la fila con CountryCode nulo
print("--- Filas con CountryCode nulo ---")
df[df['CountryCode'].isnull()]

print("\n--- Filas con CountryCode nulo (primeras 10) ---")
print(df[df['answer'].isnull()].head(10))

print("\n--- Filas con percentage nulo (primeras 10) ---")
print(df[df['percentage'].isnull()].head(10))

# Eliminar filas con CountryCode nulo
df = df.dropna(subset=['CountryCode'])

# Ahora sí
print(f"Países únicos: {df['CountryCode'].nunique()}")
print(sorted(df['CountryCode'].unique()))
```

```python
# Esto te dice todo lo que necesitas saber del dataset
print("--- Años disponibles ---")
print(df['year'].value_counts())

print("\n--- Países únicos ---")
print(sorted(df['CountryCode'].dropna().unique()))

print("\n--- Categorías disponibles ---")
print(df['category'].unique())

print("\n--- Primeras preguntas del dataset ---")
print(df['question_code'].unique()[:20])
```

```python
# Eliminar filas que no son países reales
df = df[~df['CountryCode'].isin(['Average', 'EU-28'])]

# Unificar Czech Republic y Czechia
df['CountryCode'] = df['CountryCode'].replace({'Czech Republic': 'Czechia'})

print(f"Países únicos tras limpieza: {df['CountryCode'].nunique()}")
print(sorted(df['CountryCode'].unique()))
```
